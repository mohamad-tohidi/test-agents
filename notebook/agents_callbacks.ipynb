```python
from smolagents import CodeAgent, tool, WebSearchTool
```

```python


web_search = WebSearchTool()
```

```python
_NOTES = []


@tool
def notepad(to_be_noted:str, source: str) -> str:
    """
    this is your notepad.
    whatever you save here will be availible through the process.
    if you found any hint, that worth noting. use this tool.
    

    Args:
        to_be_noted: the text you want to keep.
        source: the source that you found it from.
    Returns:
        nothing. 
    """
    _NOTES.append({"note": to_be_noted, "source": source})
    return "noted"


@tool
def get_notes() -> list:
    """
    this tool will show you the llist of notes availible.
    you must always check it before writing the full answer.
    Args:
        ...
    Returns:
        list of notes.
    """
    return _NOTES
```

```python
from smolagents.models import LiteLLMModel


model = LiteLLMModel(
    model_id="openai/gemma4:e2b-it-q4_K_M",
    api_base="http://0.0.0.0:11434/v1",
    api_key="OPENAI_API_KEY",
)
```

```python
from smolagents.monitoring import LogLevel



agent = CodeAgent(
    tools=[web_search,notepad, get_notes],
    verbosity_level = LogLevel.DEBUG,
    # step_callbacks=[scratchpad_callback],
    model=model, max_steps=3, instructions="""You are a search worker.

Your only job is to collect structured evidence for the user's question.
Do not answer the user directly.
Do not write conversational filler.
Do not make up facts.
Only use the provided tool.
Your output will be the notes that you take.
          
Rules:
- Use only the available tools: web_search, notepad, get_notes
- Search page 1 first
- If needed, continue to later pages until results are enough
- after each web search, take notes if you found a hint about the subject.
- each hint must be self contained. dont cramp multiple things into one note.
- Keep it factual and compact.
- dont miss details when writing notes. 
- always include source when taking notes.
""")
```

```python
agent.run(task="""

Find Information about Dr Reza Zia Tohidi.
          
          """)
```
