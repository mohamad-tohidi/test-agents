```python
from smolagents import CodeAgent, FinalAnswerTool, tool
```

```python
@tool
def web_search (query: str , page: int = 1) -> list[dict]:
    """
    this is the tool to search with.
    input is the query.
    page is the page, starting at page 1.
    returns a list of dict, with id and content.
    Args:
        query: The search text to look up in the DB.
        page: The results page number to return.
    Returns:
        A list of matching records as dictionaries.
    """
    return [{"id": 1, "content": "tohidi is a good boy"}, {"id":2, "content": "tohidi is a bad boy."}]
```

```python
from smolagents.models import LiteLLMModel


model = LiteLLMModel(
    model_id="openai/gemma4:e2b-it-q4_K_M",
    api_base="http://0.0.0.0:11434/v1",
    api_key="OPENAI_API_KEY",
)
```

```python
from smolagents.monitoring import LogLevel


agent = CodeAgent(
    tools=[web_search],
    verbosity_level = LogLevel.DEBUG,
    model=model, max_steps=2)
```

```python
agent.run(task="find out about mr Tohidi")
```
